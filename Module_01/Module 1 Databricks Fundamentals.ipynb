## 1. What is Data?

### Definition
**Data** is a collection of facts, figures, observations, or measurements that can be processed to extract meaningful information.

---

### 📋 Types of Data:

#### **1. Structured Data**
* **Definition:** Highly organized data with a fixed schema
* **Format:** Tables with rows and columns
* **Examples:** 
  - Relational databases (MySQL, PostgreSQL, Oracle)
  - CSV files with consistent columns
  - Spreadsheets
* **Characteristics:**
  - Easy to search and query
  - Predefined schema
  - Organized in tables
* **✅ Pros:**
  - Fast queries with SQL
  - Data integrity through constraints
  - Easy to analyze
* **❌ Cons:**
  - Rigid schema (hard to change)
  - Limited flexibility
  - Cannot handle complex nested data

**Example:**
```
Customer Table:
+----+----------+-------------------+----------+
| ID | Name     | Email             | Country  |
+----+----------+-------------------+----------+
| 1  | John Doe | john@example.com  | USA      |
| 2  | Jane Smith| jane@example.com | Canada   |
+----+----------+-------------------+----------+
```

---

#### **2. Semi-Structured Data**
* **Definition:** Data with some organizational properties but no rigid schema
* **Format:** JSON, XML, Parquet, Avro
* **Examples:**
  - JSON documents
  - XML files
  - Log files
  - Email messages
* **Characteristics:**
  - Flexible schema
  - Self-describing (contains metadata)
  - Nested structures possible
* **✅ Pros:**
  - Schema flexibility
  - Can evolve over time
  - Supports nested/hierarchical data
* **❌ Cons:**
  - More storage space
  - Slower queries than structured
  - Requires parsing

**Example (JSON):**
```json
{
  "customer_id": 1,
  "name": "John Doe",
  "orders": [
    {"order_id": 101, "amount": 50.00},
    {"order_id": 102, "amount": 75.00}
  ]
}
```

---

#### **3. Unstructured Data**
* **Definition:** Data with no predefined structure or organization
* **Format:** Text, images, audio, video
* **Examples:**
  - Documents (Word, PDF)
  - Images and videos
  - Audio files
  - Social media posts
  - Email body text
* **Characteristics:**
  - No schema
  - Requires specialized processing (NLP, computer vision)
  - Makes up ~80% of enterprise data
* **✅ Pros:**
  - Rich information content
  - Natural format for humans
  - Captures complex information
* **❌ Cons:**
  - Difficult to query
  - Requires AI/ML to extract insights
  - Large storage requirements

---

## 2. OLTP vs OLAP

### Understanding Two Fundamentally Different Workload Types

---

## 📋 OLTP (Online Transaction Processing)

### Definition:
OLTP systems handle **day-to-day transactional operations** - recording business events as they happen.

### Characteristics:
* **Purpose:** Capture and store transactions
* **Operations:** INSERT, UPDATE, DELETE (writes)
* **Query Pattern:** Short, simple queries affecting few records
* **Users:** Large number of concurrent users
* **Response Time:** Milliseconds (must be fast!)
* **Data Volume:** Current/recent data only
* **Schema:** Normalized (3NF) to avoid redundancy

### Examples:
* E-commerce order processing
* Banking transactions (deposits, withdrawals)
* Flight booking systems
* Inventory management
* Point-of-sale (POS) systems

### ✅ Pros:
* **Fast writes** - Optimized for inserts/updates
* **Data integrity** - ACID compliance
* **Concurrent access** - Handles many users
* **Real-time** - Immediate data capture

### ❌ Cons:
* **Poor for analytics** - Normalized schema slows complex queries
* **Limited history** - Usually keeps recent data only
* **Not scalable for reporting** - Queries can impact transaction performance

### Example Query:
```sql
-- Simple OLTP query: Get customer order
SELECT * FROM orders 
WHERE order_id = 12345;
```

---

## 📊 OLAP (Online Analytical Processing)

### Definition:
OLAP systems are designed for **complex analytical queries** - answering business questions from historical data.

### Characteristics:
* **Purpose:** Analyze and report on data
* **Operations:** SELECT (reads) with aggregations
* **Query Pattern:** Complex queries scanning millions of rows
* **Users:** Smaller number of analysts/business users
* **Response Time:** Seconds to minutes
* **Data Volume:** Historical data (years of history)
* **Schema:** Denormalized (star/snowflake) for query performance

### Examples:
* Sales trend analysis
* Customer behavior analytics
* Financial reporting and forecasting
* Marketing campaign performance
* Executive dashboards

### ✅ Pros:
* **Optimized for reads** - Fast complex queries
* **Historical analysis** - Years of data available
* **Aggregations** - Efficient SUM, AVG, COUNT operations
* **Business insights** - Supports decision-making

### ❌ Cons:
* **Slow writes** - Not designed for real-time inserts
* **Data latency** - Updated via periodic ETL (not real-time)
* **Storage overhead** - Denormalized = data duplication

### Example Query:
```sql
-- Complex OLAP query: Sales trend analysis
SELECT 
    region,
    product_category,
    YEAR(order_date) as year,
    QUARTER(order_date) as quarter,
    SUM(revenue) as total_revenue,
    AVG(revenue) as avg_revenue,
    COUNT(DISTINCT customer_id) as unique_customers
FROM sales_fact
JOIN date_dim ON sales_fact.date_key = date_dim.date_key
JOIN product_dim ON sales_fact.product_key = product_dim.product_key
JOIN geography_dim ON sales_fact.geo_key = geography_dim.geo_key
WHERE year >= 2022
GROUP BY region, product_category, year, quarter
ORDER BY year DESC, quarter DESC, total_revenue DESC;
```

---

## ⚖️ OLTP vs OLAP Comparison:

| Aspect | OLTP | OLAP |
|--------|------|------|
| **Purpose** | Run the business | Analyze the business |
| **Primary Operations** | INSERT, UPDATE, DELETE | SELECT with aggregations |
| **Query Complexity** | Simple | Complex |
| **Query Type** | Few rows | Millions of rows |
| **Response Time** | Milliseconds | Seconds to minutes |
| **Users** | Thousands (operational) | Hundreds (analysts) |
| **Data** | Current, detailed | Historical, summarized |
| **Schema** | Normalized (3NF) | Denormalized (Star/Snowflake) |
| **Database Size** | GB to TB | TB to PB |
| **Examples** | MySQL, PostgreSQL, Oracle | Snowflake, Redshift, BigQuery |
| **Metric** | Throughput (transactions/sec) | Query performance (time to insights) |

---

## 🔄 The Problem:

**OLTP and OLAP have conflicting requirements!**

* **OLTP needs:** Fast writes, normalized schema, ACID transactions
* **OLAP needs:** Fast reads, denormalized schema, historical data

**⚠️ You cannot optimize a database for both simultaneously!**

This fundamental conflict is why we need **separate systems**:
* **Databases** for OLTP (transactions)
* **Data Warehouses** for OLAP (analytics)

But having separate systems creates new problems (data silos, complexity, cost)...

---

### Key Insight:
The tension between OLTP and OLAP requirements is the foundational reason why data architectures evolved beyond simple databases!

## 3. Database (OLTP Systems)

### Definition:
A **database** is an organized collection of structured data stored electronically, optimized for **transactional workloads** (OLTP).

---

## 🏛️ Traditional Relational Database (RDBMS)

### Architecture:
```
┌────────────────────────────────────────┐
│       Application Layer                │
│   (CRM, ERP, Web Apps)                 │
└──────────────┬─────────────────────────┘
               │
               │ SQL Queries
               │
┌──────────────│─────────────────────────┐
│       DATABASE ENGINE                   │
│  ┌──────────────────────────────┐      │
│  │   Query Optimizer            │      │
│  └──────────────────────────────┘      │
│  ┌──────────────────────────────┐      │
│  │   Transaction Manager        │      │
│  │   (ACID Compliance)          │      │
│  └──────────────────────────────┘      │
│  ┌──────────────────────────────┐      │
│  │   Storage Engine             │      │
│  │   (B-Tree indexes)           │      │
│  └──────────────────────────────┘      │
└─────────────────────────────────────────┘
               │
               │ Disk I/O
               │
┌──────────────│─────────────────────────┐
│       DATA STORAGE                      │
│   Tables | Indexes | Logs               │
└─────────────────────────────────────────┘
```

---

### Key Features:

#### 1. **ACID Transactions**
* **Atomicity:** All or nothing (transaction succeeds completely or fails completely)
* **Consistency:** Data integrity rules are enforced
* **Isolation:** Concurrent transactions don't interfere
* **Durability:** Committed data persists even after crashes

#### 2. **Normalized Schema (3NF)**
* Data is organized to minimize redundancy
* Tables are linked via foreign keys
* Reduces storage but requires JOINs for queries

#### 3. **SQL Interface**
* Structured Query Language for data manipulation
* Declarative: describe WHAT you want, not HOW to get it

#### 4. **Indexes**
* B-Tree indexes for fast lookups
* Speed up WHERE clause filtering
* Trade-off: slower writes, faster reads

---

### ✅ Pros:

1. **ACID Compliance**
   - Strong data consistency guarantees
   - Safe for financial transactions, inventory

2. **Mature Ecosystem**
   - Decades of tooling and expertise
   - Well-understood patterns

3. **Fast Transactional Operations**
   - Optimized for single-row operations
   - Sub-millisecond response times

4. **Data Integrity**
   - Foreign keys, constraints, triggers
   - Enforces business rules at database level

5. **Concurrent Access**
   - Handles thousands of simultaneous users
   - Row-level locking

---

### ❌ Cons:

1. **Poor Analytical Performance**
   - Normalized schema requires many JOINs
   - Complex aggregations are slow
   - Queries can block transactions

2. **Limited Scalability**
   - Vertical scaling only (bigger servers)
   - Hard to distribute across multiple machines
   - Single point of failure

3. **Structured Data Only**
   - Cannot handle JSON, images, videos natively
   - Schema changes are expensive

4. **No Historical Data**
   - Usually stores current state only
   - Historical analysis requires separate archiving

5. **Cost at Scale**
   - Enterprise licenses (Oracle, SQL Server) are expensive
   - Storage and compute cannot be scaled independently

---

### Examples:

| Database | Vendor | Common Use Cases |
|----------|--------|-----------------|
| **Oracle** | Oracle Corp | Enterprise ERP, banking |
| **MySQL** | Oracle (open source) | Web applications, e-commerce |
| **PostgreSQL** | Open source | General purpose, geospatial |
| **SQL Server** | Microsoft | Windows enterprise apps |
| **DB2** | IBM | Mainframe applications |

---

### Real-World Example:

**E-Commerce Application:**
```sql
-- Insert new order (OLTP operation)
BEGIN TRANSACTION;

INSERT INTO orders (customer_id, order_date, total_amount)
VALUES (12345, '2026-08-11', 150.00);

INSERT INTO order_items (order_id, product_id, quantity, price)
VALUES (98765, 5001, 2, 50.00),
       (98765, 5002, 1, 50.00);

UPDATE inventory
SET quantity = quantity - 2
WHERE product_id = 5001;

COMMIT;
```

**What happens:**
* All operations succeed together or fail together (Atomicity)
* Inventory is updated atomically with order creation
* Fast execution (milliseconds)
* But... cannot easily answer: "What were our top-selling products last quarter?"

---

### The Problem:

Databases are **great for running the business** but **terrible for analyzing the business**.

**Business users ask:**
* "What were sales trends over the past 3 years?"
* "Which customer segments are most profitable?"
* "How does our performance compare across regions?"

**⚠️ Running these queries on the OLTP database:**
* Scans millions of rows
* Slows down customer-facing transactions
* May take minutes or time out
* Can crash the production system!

**Solution:** We need a separate system optimized for analytics → **Data Warehouse**

## 4. Data Warehouse (OLAP Systems)

### Definition:
A **Data Warehouse** is a centralized repository optimized for **analytical workloads** (OLAP) - designed for business intelligence, reporting, and data analysis.

---

## 🏗️ Data Warehouse Architecture:

```
┌────────────────────────────────────────┐
│         DATA SOURCES                     │
│   (OLTP Databases, APIs, Files)         │
└─────────────────┬───────────────────────┘
                │
       ┌────────┴────────┐
       │  ETL PROCESS   │
       │ Extract       │
       │ Transform     │  ← Nightly/hourly batch jobs
       │ Load          │
       └────────┬────────┘
                │
┌────────────────┴───────────────────────┐
│       DATA WAREHOUSE                     │
```

### Key Concepts:

#### 1. **Star Schema / Snowflake Schema**
* **Fact Tables:** Measure business events (sales, orders, clicks)
  - Contain metrics (revenue, quantity, duration)
  - Foreign keys to dimension tables
* **Dimension Tables:** Context for facts (who, what, when, where)
  - Date, product, customer, geography
  - Denormalized for fast queries

**Example Star Schema:**
```
         Customer_Dim
               |
               |
 Date_Dim ---- SALES_FACT ---- Product_Dim
               |
               |
         Geography_Dim
```

#### 2. **ETL (Extract, Transform, Load)**
* **Extract:** Pull data from source systems
* **Transform:** Clean, aggregate, join data
* **Load:** Insert into warehouse
* **Schedule:** Runs nightly/hourly (NOT real-time)

#### 3. **Columnar Storage**
* Modern warehouses store data by column, not row
* Faster aggregations (SUM, AVG, COUNT)
* Better compression

#### 4. **MPP (Massively Parallel Processing)**
* Queries distributed across multiple nodes
* Horizontal scalability
* Examples: Snowflake, Redshift, BigQuery

---

### ✅ Pros:

1. **Optimized for Analytics**
   - Fast complex queries with JOINs
   - Efficient aggregations (SUM, AVG, GROUP BY)
   - Sub-second to seconds response time

2. **Historical Data**
   - Years of history for trend analysis
   - Slowly Changing Dimensions (SCD) support

3. **Business Intelligence**
   - Seamless integration with BI tools
   - Standardized reporting layer
   - Consistent metrics across organization

4. **Separate from OLTP**
   - Analytics don't impact production systems
   - Can run large queries without fear

5. **Optimized Schema**
   - Denormalized for query performance
   - Pre-aggregated summaries

---

### ❌ Cons:

1. **Structured Data Only**
   - Cannot handle JSON, images, videos, logs
   - Must fit into predefined schema
   - Schema changes are expensive and slow

2. **Data Latency**
   - ETL runs periodically (hours to days)
   - Not real-time or near-real-time
   - "What happened yesterday" not "What's happening now"

3. **High Cost**
   - Storage costs for historical data
   - Compute costs for queries
   - Licensing (Oracle, Teradata)

4. **Proprietary Formats**
   - Data locked in vendor-specific format
   - Difficult to share data externally
   - Vendor lock-in

5. **Duplicate Data**
   - Data copied from OLTP to warehouse
   - Storage duplication
   - Synchronization overhead

6. **Poor for Machine Learning**
   - Cannot store training data (images, text, embeddings)
   - Limited support for unstructured data
   - Expensive for large-scale ML

## Key Components:

### 1. **HDFS (Hadoop Distributed File System)**
* **Distributed storage** across commodity hardware
* **Replication:** Data stored 3x for fault tolerance
* **Block-based:** Files split into 64MB/128MB blocks
* **Scalable:** Add more nodes = more storage

### 2. **MapReduce (Programming Model)**
* **Map phase:** Process data in parallel
* **Reduce phase:** Aggregate results
* **Automatic parallelization:** Framework handles distribution
* **Fault tolerance:** Restarts failed tasks

---

## How MapReduce Works:

**Example: Word Count**

```
Input: Large text files

1. MAP PHASE (Parallel)
   Node 1: "hello world" → [(hello, 1), (world, 1)]
   Node 2: "hello spark" → [(hello, 1), (spark, 1)]
   Node 3: "world data" → [(world, 1), (data, 1)]

2. SHUFFLE PHASE
   Group by key:
   hello: [1, 1]
   world: [1, 1]
   spark: [1]
   data: [1]

3. REDUCE PHASE (Parallel)
   hello: sum([1, 1]) = 2
   world: sum([1, 1]) = 2
   spark: sum([1]) = 1
   data: sum([1]) = 1

Output: {hello: 2, world: 2, spark: 1, data: 1}
```

## ✅ Hadoop Revolution - What It Got Right:

1. **Horizontal Scalability**
   - Add more machines, not bigger machines
   - Linear scalability to 1000s of nodes

2. **Cost-Effective**
   - Commodity hardware ($1,000 vs $50,000 per node)
   - Open source (free software)
   - 10-100x cheaper than data warehouses

3. **Fault Tolerance**
   - Automatic retry of failed tasks
   - Data replication (no data loss)

4. **All Data Types**
   - Store structured, semi-structured, unstructured
   - No schema required

5. **Proven at Scale**
   - Yahoo, Facebook processing petabytes
   - Industry validation

**Impact:** Hadoop enabled companies like Yahoo, Facebook, and LinkedIn to process petabytes of data!

## ❌ Why Hadoop/MapReduce Failed to Dominate:

### **Critical Problems:**

### 1. **Too Slow**
* **Disk-based:** Every operation writes to disk
* **Batch-only:** No real-time or interactive queries
* **Minutes to hours:** Even simple queries take forever
* **Example:** A simple aggregation that takes 2 seconds in SQL takes 5 minutes in MapReduce!

### 2. **Too Hard to Use**
* **Low-level programming:** Write Map and Reduce functions manually
* **No SQL support:** Can't use familiar SQL
* **Requires Java expertise:** Not accessible to analysts
* **Complex debugging:** Distributed failures are hard to diagnose

**Example MapReduce Code (Just for Word Count!):**
```java
public class WordCount {
  public static class Map extends Mapper<LongWritable, Text, Text, IntWritable> {
    public void map(LongWritable key, Text value, Context context) {
      String line = value.toString();
      String[] words = line.split(" ");
      for (String word : words) {
        context.write(new Text(word), new IntWritable(1));
      }
    }
  }
  
  public static class Reduce extends Reducer<Text, IntWritable, Text, IntWritable> {
    public void reduce(Text key, Iterable<IntWritable> values, Context context) {
      int sum = 0;
      for (IntWritable val : values) {
        sum += val.get();
      }
      context.write(key, new IntWritable(sum));
    }
  }
}
```

**vs SQL (Simple!):**
```sql
SELECT word, COUNT(*) FROM text GROUP BY word;
```

### 3. **No Iterative Processing**
* **Machine learning requires iteration:** Train for 100s of epochs
* **MapReduce restarts from disk each time:** 100x slower than needed
* **Can't keep data in memory**

### 4. **Poor for Analytics**
* **No interactive queries:** Can't explore data
* **No ACID transactions:** Data quality issues
* **No schema enforcement:** Data swamps
* **BI tools can't connect:** Requires custom code

### 5. **Operational Complexity**
* **Complex to deploy and manage**
* **Many moving parts** (NameNode, DataNode, JobTracker, TaskTracker)
* **Requires dedicated DevOps team**
* **Version compatibility issues**

**Unified analytics engine** for large-scale data processing with:
* **In-memory processing**
* **High-level APIs** (Python, SQL, Scala, R)
* **Unified platform** (batch, streaming, ML, graph)

---

## 🏆 How Spark Solved Hadoop's Problems:

### **Problem 1: Hadoop Too Slow**
**Spark Solution: In-Memory Processing**
* **100x faster** for in-memory operations
* **10x faster** for disk operations
* Keeps data in RAM between operations
* No constant disk I/O

**Comparison:**
```
Same Query:
- MapReduce (disk): 5 minutes
- Spark (memory): 3 seconds

= 100x FASTER!
```

---

### **Problem 2: MapReduce Too Hard**
**Spark Solution: High-Level APIs**

**Word Count in Spark (Python):**
```python
# Just 2 lines!
df = spark.read.text("file.txt")
df.selectExpr("explode(split(value, ' ')) as word") \
  .groupBy("word").count().show()
```

**Or even simpler with SQL:**
```sql
SELECT word, COUNT(*) 
FROM (SELECT explode(split(value, ' ')) as word FROM text)
GROUP BY word;
```

**vs MapReduce: 50+ lines of Java!**

---

### **Problem 3: No Iterative Processing (ML)**
**Spark Solution: RDD Caching**
* Cache datasets in memory
* Reuse across iterations
* **100x faster** for machine learning

**Example: ML Training**
```
MapReduce:
- Iteration 1: Load from disk (slow)
- Iteration 2: Load from disk AGAIN (slow)
- Iteration 100: Still loading from disk!

Spark:
- Iteration 1: Load from disk, cache in memory
- Iterations 2-100: Use cached data (fast!)
```

---

### **Problem 4: Batch-Only**
**Spark Solution: Unified Engine**
* **Spark SQL** - Interactive queries
* **Spark Streaming** - Real-time processing
* **Spark MLlib** - Machine learning
* **Spark GraphX** - Graph processing

**Same engine, different workloads!**

---

### **Problem 5: Complexity**
**Spark Solution: Simpler Architecture**

**Hadoop:** NameNode, DataNode, JobTracker, TaskTracker, ResourceManager, NodeManager...

**Spark:** Driver + Executors. That's it!

* **RDDs:** Low-level control when needed

### 4. **Multi-Language Support**
* Python (PySpark)
* Scala (native)
* Java
* R (SparkR)
* SQL

## 5. Data Lake

### Definition:
A **Data Lake** is a centralized repository that stores **ALL data** (structured, semi-structured, unstructured) in its **raw, native format** at massive scale.

---

## 🌊 Data Lake Architecture:

```
┌────────────────────────────────────────┐
│         DATA SOURCES                     │
│   Databases | Logs | IoT | Social Media │
│   APIs | Files | Images | Videos         │
└────────────────┬───────────────────────┘
                │
       ┌────────┴────────┐
       │ INGESTION     │
       │ (ELT not ETL) │  ← Load FIRST, transform LATER
       └────────┬────────┘
                │
┌────────────────┴───────────────────────┐
│       DATA LAKE                          │
│  (Cloud Object Storage: S3, ADLS, GCS) │
│                                          │
│  ┌─────────────────────────────┐  │
│  │ RAW ZONE                  │  │
│  │ (Original files as-is)    │  │
│  └─────────────────────────────┘  │
│  ┌─────────────────────────────┐  │
│  │ PROCESSED ZONE            │  │
│  │ (Cleaned, Parquet)        │  │
│  └─────────────────────────────┘  │
│  ┌─────────────────────────────┐  │
│  │ CURATED ZONE              │  │
│  │ (Analytics-ready)         │  │
│  └─────────────────────────────┘  │
└────────────────┬───────────────────────┘
                │
┌────────────────┴───────────────────────┐
│         PROCESSING                       │
│  Spark | Hadoop | Python | ML           │
└───────────────────────────────────────┘
```

---

### Key Characteristics:

#### 1. **Schema-on-Read**
* Store data WITHOUT defining schema upfront
* Schema applied when data is READ (not when written)
* Flexible: add new data types anytime

#### 2. **All Data Types**
* **Structured:** CSV, Parquet, Avro
* **Semi-structured:** JSON, XML, logs
* **Unstructured:** Images, videos, PDFs, audio

#### 3. **Cost-Effective Storage**
* Uses cheap cloud object storage (S3, ADLS, GCS)
* Pay only for what you store
* Pennies per GB/month

#### 4. **Big Data Processing**
* Apache Spark for distributed processing
* Hadoop ecosystem tools
* Machine learning frameworks

#### 5. **ELT Instead of ETL**
* **Extract:** Pull data from sources
* **Load:** Store raw data immediately
* **Transform:** Process later as needed

---

### ✅ Pros:

1. **Handles All Data Types**
   - Structured, semi-structured, unstructured
   - No schema constraints
   - Store anything

2. **Cost-Effective**
   - Cheap cloud storage ($0.02/GB/month on S3)
   - Store ALL data, not just what's "important"
   - Decouple storage from compute

3. **Scalability**
   - Store petabytes of data
   - Horizontal scaling (add more storage)
   - No database size limits

4. **Flexibility**
   - Schema-on-read (define schema later)
   - Support for ML/AI workloads
   - Data science friendly

5. **Future-Proof**
   - Keep raw data for future use cases
   - "Store now, ask questions later"

---

### ❌ Cons:

1. **No ACID Transactions**
   - Cannot update or delete records reliably
   - No transactional consistency
   - Eventual consistency only

2. **Poor Query Performance**
   - Full file scans required
   - No indexes
   - Slow for point queries

3. **Data Quality Issues**
   - No schema enforcement
   - Easy to become a "data swamp"
   - Requires good governance

4. **Complex Tooling**
   - Requires Spark/Hadoop expertise
   - Steep learning curve
   - DevOps overhead

5. **No Built-in BI Support**
   - Cannot connect Tableau/Power BI directly
   - Must create processed tables first
   - Extra ETL step needed

6. **Data Reliability**
   - Files can be partially written (no atomicity)
   - Difficult to maintain consistency
   - Hard to implement SLAs

---

**Clickstream Analysis with Spark:**

```python
df = spark.read.json("s3://my-lake/raw/clickstream/*.json")

# Process and aggregate
user_sessions = df \
    .groupBy("user_id", "session_id") \
    .agg(
        count("*").alias("page_views"),
        sum("time_spent").alias("total_time"),
        collect_set("page_url").alias("pages_visited")
    )

# Write to processed zone as Parquet
user_sessions.write \
    .mode("append") \
    .partitionBy("date") \
    .parquet("s3://my-lake/processed/user_sessions/")
```

## Cost Considerations:

**Pricing Model:**
```
Total Cost = Cloud Infrastructure Cost + Databricks Platform Cost (DBUs)
```

**DBU (Databricks Unit):**
* Normalized unit of processing capability
* Varies by:
  - Edition (Standard/Premium/Enterprise)
  - Workload type (All-Purpose, Jobs, SQL, ML)
  - Cloud provider
  - Region

**Cost Optimization Tips:**
1. Use **Job Clusters** instead of All-Purpose for production
2. Enable **Auto-termination** for idle clusters
3. Use **Spot/Preemptible instances** for fault-tolerant workloads
4. Right-size clusters (avoid over-provisioning)
5. Use **Serverless** for variable workloads
6. Monitor and optimize with **Cost Management** tools
