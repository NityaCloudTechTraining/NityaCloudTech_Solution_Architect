```python
json_data={
    "Id":123,
    "Name":"Amit",
    "Address":["123 Main Street","New York","NY"],
    "Contact":{"Phone":"123-456-7890",
                "Email":"abc@def.com"}
}
```

```python
df=spark.createDataFrame([json_data])

df.printSchema()
```

```python
schema=df.schema
```

```python
for field in schema.fields:
    print(field.name, " ---> ", field.dataType)


```

```python
from pyspark.sql.functions import explode_outer,explode
df2=df.withColumn("Address",explode("Address")).select("Contact.Phone","Address")

display(df2)
```

```python
json_data = {
    "Id": 123,
    "Name": "Amit",
    "Address": [
        {
            "Street": "123 Main Street",
            "City": "New York",
            "State": "NY"
        },
        {
            "Street": "456 Park Avenue",
            "City": "Chicago",
            "State": "IL"
        }
    ],
    "Contact": {
        "Phone": "123-456-7890",
        "Email": "abc@def.com"
    }
}
```

```python
df=spark.createDataFrame([json_data])
```

```python
df.printSchema()
```

```python
from pyspark.sql.functions import explode

df2 = df.withColumn(
    "Address",
    explode("Address")
)

display(df2)
```

```python
df3 = df2.select(
    "Id",
    "Name",
    "Address.Street",
    "Address.City",
    "Address.State",
    "Contact.Phone",
    "Contact.Email"
)
```

```python
df3.show()
```
