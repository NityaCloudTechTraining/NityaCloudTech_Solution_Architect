```python
# What is dbutils?

# dbutils is a Databricks utility package that provides various helper functions
# for working with files, notebooks, secrets, and widgets in Databricks.

# Key modules:
# - dbutils.fs: File system utilities for DBFS, S3, Azure Blob Storage, etc.
# - dbutils.notebook: Run and manage notebooks programmatically
# - dbutils.secrets: Securely access stored credentials
# - dbutils.widgets: Create interactive input widgets

# Let's explore the most commonly used dbutils.fs commands!

print("Available dbutils modules:")
print("- dbutils.fs: File system operations")
print("- dbutils.notebook: Notebook workflows")
print("- dbutils.secrets: Secret management")
print("- dbutils.widgets: UI input widgets")
print("\nToday we'll focus on dbutils.fs commands for file operations!")
```

```python
# dbutils.fs.ls() - List Files and Directories
# Purpose: Lists all files and directories in a specified path
# Returns: List of FileInfo objects with name, path, size, and isDir properties

# Example 1: List contents of your S3 path
print("=== Listing contents of s3://nitya-cloutech/Account/HR/ ===")
try:
    files = dbutils.fs.ls("s3://nitya-cloutech/Account/HR/")
    
    for file in files:
        file_type = "DIR" if file.isDir() else "FILE"
        size_mb = file.size / (1024 * 1024)  # Convert to MB
        print(f"{file_type:6} | {file.name:40} | Size: {size_mb:.2f} MB")
        
except Exception as e:
    print(f"Error: {e}")
    print("Note: Make sure the path exists and you have proper AWS credentials configured")

# Example 2: You can also use display() for better visualization
# display(dbutils.fs.ls("s3://nitya-cloutech/Account/HR/"))
```

```python
# dbutils.fs.mkdirs() - Create Directories
# Purpose: Creates a directory and all necessary parent directories
# Returns: True if successful

# Example 1: Create a new subdirectory in your S3 path
print("=== Creating new directories ===")

try:
    # Create a new folder for employee data
    new_dir = "s3://nitya-cloutech/Account/HR/EmployeeData/"
    result = dbutils.fs.mkdirs(new_dir)
    print(f"✓ Directory created: {new_dir}")
    print(f"Result: {result}")
    
    # Create nested directories (all parents will be created automatically)
    nested_dir = "s3://nitya-cloutech/Account/HR/Reports/2026/August/"
    result = dbutils.fs.mkdirs(nested_dir)
    print(f"✓ Nested directory created: {nested_dir}")
    print(f"Result: {result}")
    
    print("\n✓ All directories created successfully!")
    
except Exception as e:
    print(f"Error: {e}")
    print("Note: Ensure you have write permissions to the S3 bucket")
```

```python
# dbutils.fs.cp() - Copy Files and Directories
# Purpose: Copies files or directories from source to destination
# Syntax: dbutils.fs.cp(source, destination, recurse=False)
# Parameters:
#   - source: Source file or directory path
#   - destination: Destination path
#   - recurse: If True, copies directories recursively

print("=== Copying files and directories ===")

try:
    # Example 1: Copy a single file
    # First, let's create a sample file to copy
    sample_file = "s3://nitya-cloutech/Account/HR/sample.txt"
    dbutils.fs.put(sample_file, "This is a sample HR document", overwrite=True)
    print(f"✓ Created sample file: {sample_file}")
    
    # Copy the file to a backup location
    source = "s3://nitya-cloutech/Account/HR/sample.txt"
    destination = "s3://nitya-cloutech/Account/HR/EmployeeData/sample_backup.txt"
    dbutils.fs.cp(source, destination)
    print(f"✓ Copied file: {source} -> {destination}")
    
    # Example 2: Copy entire directory (with recurse=True)
    # dbutils.fs.cp("s3://nitya-cloutech/Account/HR/Reports/", 
    #               "s3://nitya-cloutech/Account/HR/Reports_Backup/", 
    #               recurse=True)
    # print("✓ Directory copied recursively")
    
    print("\n✓ Copy operations completed!")
    
except Exception as e:
    print(f"Error: {e}")
```

```python
# dbutils.fs.rm() - Remove/Delete Files and Directories
# Purpose: Deletes files or directories
# Syntax: dbutils.fs.rm(path, recurse=False)
# Parameters:
#   - path: Path to file or directory to delete
#   - recurse: If True, deletes directories and all contents recursively
# WARNING: This operation is permanent! Use with caution.

print("=== Removing files and directories ===")

try:
    # Example 1: Remove a single file
    file_to_remove = "s3://nitya-cloutech/Account/HR/sample.txt"
    
    # Check if file exists before removing
    try:
        dbutils.fs.ls(file_to_remove)
        dbutils.fs.rm(file_to_remove)
        print(f"✓ Removed file: {file_to_remove}")
    except:
        print(f"File not found or already deleted: {file_to_remove}")
    
    # Example 2: Remove a directory and all its contents (recurse=True)
    # CAUTION: This will delete everything inside!
    # dir_to_remove = "s3://nitya-cloutech/Account/HR/EmployeeData/"
    # dbutils.fs.rm(dir_to_remove, recurse=True)
    # print(f"✓ Removed directory recursively: {dir_to_remove}")
    
    # Example 3: Safe deletion - check before removing
    path_to_check = "s3://nitya-cloutech/Account/HR/test_folder/"
    try:
        files = dbutils.fs.ls(path_to_check)
        print(f"⚠ Warning: {path_to_check} contains {len(files)} items")
        print("Uncomment the rm command to delete it")
        # dbutils.fs.rm(path_to_check, recurse=True)
    except:
        print(f"Path does not exist: {path_to_check}")
    
    print("\n✓ Remove operations completed!")
    
except Exception as e:
    print(f"Error: {e}")
```

```python
# dbutils.fs.mv() - Move or Rename Files and Directories
# Purpose: Moves or renames files/directories
# Syntax: dbutils.fs.mv(source, destination, recurse=False)
# Note: This is essentially a rename operation

print("=== Moving/Renaming files ===")

try:
    # Example 1: Create a test file
    original_path = "s3://nitya-cloutech/Account/HR/old_name.txt"
    dbutils.fs.put(original_path, "Test content for move operation", overwrite=True)
    print(f"✓ Created test file: {original_path}")
    
    # Example 2: Rename a file (move within same directory)
    new_path = "s3://nitya-cloutech/Account/HR/new_name.txt"
    dbutils.fs.mv(original_path, new_path)
    print(f"✓ Renamed: {original_path} -> {new_path}")
    
    # Example 3: Move a file to different directory
    final_path = "s3://nitya-cloutech/Account/HR/Reports/moved_file.txt"
    dbutils.fs.mv(new_path, final_path)
    print(f"✓ Moved: {new_path} -> {final_path}")
    
    # Example 4: Move entire directory (with recurse=True)
    # dbutils.fs.mv("s3://nitya-cloutech/Account/HR/OldFolder/",
    #               "s3://nitya-cloutech/Account/HR/NewFolder/",
    #               recurse=True)
    # print("✓ Directory moved")
    
    print("\n✓ Move operations completed!")
    
except Exception as e:
    print(f"Error: {e}")
```

```python
# dbutils.fs.put() - Write Content to a File
# Purpose: Creates a file and writes string content to it
# Syntax: dbutils.fs.put(path, contents, overwrite=False)
# Parameters:
#   - path: Destination file path
#   - contents: String content to write
#   - overwrite: If True, overwrites existing file
# Note: Best for small text files. For large data, use Spark or other methods.

print("=== Writing content to files ===")

try:
    # Example 1: Create a simple text file
    file_path = "s3://nitya-cloutech/Account/HR/employee_notes.txt"
    content = """HR Department Notes
====================
Date: August 26, 2026

Employee Records:
- Total Employees: 150
- New Hires This Month: 5
- Department: Human Resources
"""
    dbutils.fs.put(file_path, content, overwrite=True)
    print(f"✓ Created file: {file_path}")
    
    # Example 2: Create a CSV file
    csv_path = "s3://nitya-cloutech/Account/HR/employee_summary.csv"
    csv_content = """EmployeeID,Name,Department,JoinDate
101,John Doe,Engineering,2024-01-15
102,Jane Smith,HR,2024-03-20
103,Bob Johnson,Sales,2024-05-10
"""
    dbutils.fs.put(csv_path, csv_content, overwrite=True)
    print(f"✓ Created CSV: {csv_path}")
    
    # Example 3: Append is not directly supported - need to read, modify, write
    # To append, you would need to:
    # 1. Read existing content with dbutils.fs.head()
    # 2. Concatenate new content
    # 3. Write back with overwrite=True
    
    print("\n✓ File write operations completed!")
    
except Exception as e:
    print(f"Error: {e}")
```

```python
# dbutils.fs.head() - Read File Content
# Purpose: Reads and returns the first N bytes of a file
# Syntax: dbutils.fs.head(path, maxBytes=65536)
# Parameters:
#   - path: File path to read
#   - maxBytes: Maximum number of bytes to read (default: 65536 = 64KB)
# Returns: String content of the file

print("=== Reading file content ===")

try:
    # Example 1: Read the text file we created earlier
    file_path = "s3://nitya-cloutech/Account/HR/employee_notes.txt"
    
    try:
        content = dbutils.fs.head(file_path)
        print(f"Content of {file_path}:")
        print("=" * 50)
        print(content)
        print("=" * 50)
    except:
        print(f"File not found: {file_path}")
        print("Run the previous cell (put command) to create this file first")
    
    # Example 2: Read first 100 bytes only
    csv_path = "s3://nitya-cloutech/Account/HR/employee_summary.csv"
    try:
        partial_content = dbutils.fs.head(csv_path, maxBytes=100)
        print(f"\nFirst 100 bytes of {csv_path}:")
        print("=" * 50)
        print(partial_content)
        print("=" * 50)
    except:
        print(f"File not found: {csv_path}")
    
    # Note: For large files or structured data processing, 
    # use Spark DataFrame methods instead:
    # df = spark.read.csv("s3://path/to/file.csv")
    
    print("\n✓ File read operations completed!")
    
except Exception as e:
    print(f"Error: {e}")
```
